```python
print("Fighter inherits from Character:",
    issubclass(Fighter, Character))

print("Mage inherits from Character",
      issubclass(Mage, Character))

print("MeleeWeapon inherits from Weapon:",
    issubclass(MeleeWeapon, Weapon))

print("RangedWeapon inherits from Weapon",
      issubclass(RangedWeapon, Weapon))

print("MagicWeapon inherits from Weapon",
      issubclass(MagicWeapon, Weapon))
```
