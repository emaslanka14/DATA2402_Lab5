## Verfiy Relationships

```python
print("Fighter inherits from character:", issubclass(Fighter, Character))
print("Mage inherits from character:", issubclass(Mage, Character))
print("MeleeWeapon inherits from character:", issubclass(MeleeWeapon, Weapon))
print("RangedWeapon inherits from weapon:", issubclass(RangedWeapon, Weapon))
print("MagicWeapon inherits from character:", issubclass(MagicWeapon, Weapon))
```

## Normal Case - Attacking as a Mage

```python
glub = Fighter("glub", 150, 50)
magicClub = MagicWeapon("Magical Club", 76, 60, 6.5)
wizardDude = Mage("wizardDude", 90, 200)

attack(wizardDude, magicClub, glub, 5)
```

## Special Case - Trying to Attack with a MagicWeapon as not a Mage

```python
try:
    attack(glub, magicClub, wizardDude,  0)
except Exception as e:
    print(e)
```
